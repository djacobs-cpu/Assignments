```python
import pandas as pd
import matplotlib.pyplot as plt


df = pd.read_csv('sales_data.csv')


sales_dict = df.to_dict(orient='records')


total_sales = df['sales'].sum()
print(f"Total sales: {total_sales}")


region_sales = df.groupby('region')['sales'].sum().sort_values(ascending=False)
top_three_regions = region_sales.head(3)
print("Top three regions by overall products sold:")
print(top_three_regions)


product_sales = df.groupby('product')['sales'].sum()
more_sold = product_sales.idxmax()
print(f"{more_sold} sold more.")


df['revenue'] = df.apply(lambda row: row['sales'] * 10.99 if row['product'] == 'bobbles' else row['sales'] * 15.99, axis=1)
product_revenue = df.groupby('product')['revenue'].sum()
most_revenue_product = product_revenue.idxmax()
print("Overall revenue for each product:")
print(product_revenue)
print(f"{most_revenue_product} made the most revenue.")


least_sales = df.loc[df['sales'].idxmin()]
print(f"The region and product that sold the least: {least_sales['region']} - {least_sales['product']}")


region_sales.plot(kind='bar')
plt.title('Overall Sales by Region')
plt.xlabel('Region')
plt.ylabel('Sales')
plt.show()


product_sales.plot(kind='pie', autopct='%1.1f%%')
plt.title('Overall Sales by Product')
plt.ylabel('')
plt.show()
```
