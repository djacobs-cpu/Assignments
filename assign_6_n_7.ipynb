```python
import csv
from collections import defaultdict

with open('sales_data.csv', mode='r') as file:
    csv_reader = csv.DictReader(file)
    sales_data = list(csv_reader)

total_sales = sum(int(row['sales']) for row in sales_data)
print(f"Total sales: {total_sales}")

region_sales = defaultdict(int)
for row in sales_data:
    region_sales[row['region']] += int(row['sales'])

sorted_regions = sorted(region_sales.items(), key=lambda x: x[1], reverse=True)
top_three_regions = sorted_regions[:3]
print("Top three regions by sales:")
for region, sales in top_three_regions:
    print(f"{region}: {sales}")

product_sales = defaultdict(int)
for row in sales_data:
    product_sales[row['product']] += int(row['sales'])

if product_sales['bobbles'] > product_sales['widgets']:
    print("Bobbles sold more than widgets.")
else:
    print("Widgets sold more than bobbles.")

bobble_price = 10.99
widget_price = 15.99

bobble_revenue = product_sales['bobbles'] * bobble_price
widget_revenue = product_sales['widgets'] * widget_price

print(f"Bobbles revenue: ${bobble_revenue:.2f}")
print(f"Widgets revenue: ${widget_revenue:.2f}")

if bobble_revenue > widget_revenue:
    print("Bobbles made more revenue.")
else:
    print("Widgets made more revenue.")

min_sale = min(int(row['sales']) for row in sales_data)
for row in sales_data:
    if int(row['sales']) == min_sale:
        print(f"Region and product with the least sales: {row['region']}, {row['product']}")
        break
```
